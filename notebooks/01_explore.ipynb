# 01 探索：从 IBKR 拉数据并看一眼
先在 TWS 登录，再运行。

```python
from ibkr_quant.ibkr_client import ib_session
from ibkr_quant.data.fetch import stock, historical_bars
from ibkr_quant.data.store import save_bars, load_closes
from ibkr_quant.analysis.returns import summary, simple_returns
```

```python
with ib_session() as ib:
    con = stock('AAPL'); ib.qualifyContracts(con)
    df = historical_bars(ib, con, duration='1 Y', bar_size='1 day')
save_bars(df, 'AAPL')
df.tail()
```

```python
closes = load_closes(['AAPL','MSFT','SPY']).dropna()
summary(closes).round(4)
```

```python
(1 + simple_returns(closes)).cumprod().plot(figsize=(10,5), grid=True, title='NAV');
```
