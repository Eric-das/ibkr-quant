# 02 拉取历史数据到本地 Parquet
改 `symbols` / `duration` / `bar` 后运行即可，重复运行会自动合并去重。

```python
from ibkr_quant.ibkr_client import ib_session
from ibkr_quant.data.fetch import stock, historical_bars
from ibkr_quant.data.store import save_bars

symbols = ['SPY', 'QQQ', 'IWM', 'TLT', 'GLD']
duration, bar = '8 Y', '1 day'
```

```python
with ib_session() as ib:
    for s in symbols:
        con = stock(s); ib.qualifyContracts(con)
        df = historical_bars(ib, con, duration=duration, bar_size=bar)
        save_bars(df, s, bar)
        print(f'{s}: {len(df)} 根  {df.index[0].date()} ~ {df.index[-1].date()}')
```
